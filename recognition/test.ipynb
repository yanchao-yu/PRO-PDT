```python
from recognition.core.inference.furhat_service import FurhatIdentitySystem
from recognition.core.utils.database import Database
from furhat_realtime_api import AsyncFurhatClient


def handle_vision_event(event_data: dict):

        print(f"[Vision] Received event: {event_data}")
        if not event_data:
            return

        event_type = event_data.get("event")
        name = event_data.get("name", "unknown")
        role = event_data.get("role")

        print(f"[VISION EVENT] Type: {event_type} | Name: {name} | Role: {role}")

furhat = AsyncFurhatClient(host="127.0.0.1")
print(furhat)
vision = FurhatIdentitySystem(
    model_name="inception_resnet_v1",
    model_path="core/weights/resnet_best.pth",
    database=Database().get_all(),
    use_webcam=True,
    callback=handle_vision_event,
    furhat_client=furhat,
)

await vision.run()
```
